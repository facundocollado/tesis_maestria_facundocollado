```python
import pandas as pd
from google.colab import drive
from google.colab import files
drive.mount('/content/drive')
# Dataset is now stored in a Pandas Dataframe
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/indeed_job_dataset.csv')
df['Description']
```

```python
#Proceso de limpieza como lo descripto en: https://medium.com/analytics-vidhya/key-phrase-extraction-applause-prediction-7b397c7ad76d
#lowercase
df['Description'] = df['Description'].str.lower()
#Clear HTML tags
df['Description'] = df['Description'].str.replace(r'<[^<>]*>', ' ', regex=True)
#remove emails
df['Description'] = df['Description'].str.replace(r'\S*@\S*\s?', '', regex=True)
#remove urls
df['Description'] = df['Description'].str.replace('http\S+|www.\S+', '', case=True)
#remove \r and \n
df['Description'] = df['Description'].str.replace(r'\r', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'\n', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'\a', ' ', regex=True)
#remove [(/)]
df['Description'] = df['Description'].str.replace(r'\[', '', regex=True)
df['Description'] = df['Description'].str.replace(r'\]', '', regex=True)

#noisy words regarding identity, sex, status religion and other non-merit based factors from the applicant and unrelated (trash) words
remove_list = {"qualified", "applicant", "receive", "consideration",
               "regard", "race", "sexual orientation", "gender identity",
               "national", "origin", "color", "religion", "sex", "disability",
               "ancestry", "religious creed", "creed",
               "political affiliation", "military service", "citizenship",
               "equal employment opportunities", "eoe", "minority", "female",
               "veteran","disabled", "protected", "marital status", "marital",
               "veteran status", "expression", "applicable", "federal",
               "union affiliation",
               "state", "local", "law", "year", "years", "experience", "hiv",
               "taxes", "biking", "trails", "surf", "food", "retirement",
               "sclerosis", "oncology", "immunology", "banking", "clinical terms",
               "holidays", "holiday", "vacations", "vacation",
               "high-deductible", "health", "HDHP", "hawaiian", "islands",
               "willamette river waterfront", "financial services", "credit card",
               "highest Presidential honor", "performance excellence",
               "chronic pain", "discomfort",
               #business names
               "credit suisse", "add systems", "foxconn", "evonik"}
for word in remove_list:
  df['Description'] = df['Description'].str.replace(word, ' ', regex=True)

df['Description'] = df['Description'].str.replace(r'\/', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'\(', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'\)', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'\;', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'…', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'_', ' ', regex=True)

df['Description'] = df['Description'].str.replace(r' k ', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r' PTO ', ' ', regex=True)

#’ to ' for apostrophe
df['Description'] = df['Description'].str.replace(r'’', "'", regex=True)
#Replace mgmt to management
df['Description'] = df['Description'].str.replace(r'mgmt', 'management', regex=True)
#remove ampersons html
df['Description'] = df['Description'].str.replace(r'&amp;', ' ', regex=True)
#Punctuation Removal -> ‘.”,?~!<@#$%&*)
list = [".", ",", "?", "~", "!", "<", ">", "@", "#", "$", "%", "&", "*", "-", "–" ,"§", "+", "·", '"', ":", "•", "—", '“', '”', '”', "®", "'"]
df["Description"] = df["Description"].str.translate({ord(x): ' ' for x in list})
df['Description'] = df['Description'].str.replace(r' s ', ' ', regex=True)
#Remove numbers
df['Description'] = df['Description'].str.replace('\d+', ' ', regex=True)
#remove multiple spaces '   ' => ' '
df['Description'] = df['Description'].replace(r'\s+', ' ', regex=True)

# Remover marcas y nombres de organizaciones
df['Description'] = df['Description'].str.replace(r'kpmg', '', regex=True)
df['Description'] = df['Description'].str.replace(r' ESI ', ' ', regex=True)
df['Description'] = df['Description'].str.replace(r'xcelsius', '', regex=True)
df['Description'] = df['Description'].str.replace(r'eeo', '', regex=True)
df['Description'] = df['Description'].str.replace(r'msja', '', regex=True)
df['Description'] = df['Description'].str.replace(r' ddi ', '', regex=True)
df['Description'] = df['Description'].str.replace(r'basf', '', regex=True)
df['Description'] = df['Description'].str.replace(r'procter', '', regex=True)
df['Description'] = df['Description'].str.replace(r'arcelormittal', '', regex=True)
df['Description'] = df['Description'].str.replace(r'wommelgem', '', regex=True)
df['Description'] = df['Description'].str.replace(r'wsj', '', regex=True)
df['Description'] = df['Description'].str.replace(r'nwslv', '', regex=True)
df['Description'] = df['Description'].str.replace(r'nwsa', '', regex=True)
df['Description'] = df['Description'].str.replace(r'nws', '', regex=True)
df['Description'] = df['Description'].str.replace(r'nasdaq', '', regex=True)
df['Description'] = df['Description'].str.replace(r'asx', '', regex=True)
df['Description'] = df['Description'].str.replace(r'vlookups', '', regex=True)
df['Description'] = df['Description'].str.replace(r'vlookup', '', regex=True)
df['Description'] = df['Description'].str.replace(r'smurfit kappa', '', regex=True)
df['Description'] = df['Description'].str.replace(r'fcra', '', regex=True)
df['Description'] = df['Description'].str.replace(r'nacha', '', regex=True)
df['Description'] = df['Description'].str.replace(r'tcpa', '', regex=True)
df['Description'] = df['Description'].str.replace(r'bsa', '', regex=True)
df['Description'] = df['Description'].str.replace(r'ofac', '', regex=True)
df['Description'] = df['Description'].str.replace(r'scifs', 'sensitive compartmented information facility', regex=True)
df['Description'] = df['Description'].str.replace(r'nga', '', regex=True)
df['Description'] = df['Description'].str.replace(r'dvcs', '', regex=True)
df['Description'] = df['Description'].str.replace(r'allegheny', '', regex=True)
df['Description'] = df['Description'].str.replace(r' anet ', '', regex=True)
df['Description'] = df['Description'].str.replace(r'deloitte', '', regex=True)
df['Description'] = df['Description'].str.replace(r' llp ', '', regex=True)
df['Description'] = df['Description'].str.replace(r'koch', '', regex=True)
df['Description'] = df['Description'].str.replace(r'ksas', '', regex=True)
df['Description'] = df['Description'].str.replace(r'ochsner', '', regex=True)
df['Description'] = df['Description'].str.replace(r'raytheon', '', regex=True)

#df.apply(lambda x: x.dropna().reset_index(drop = True))

#remove all empty Description
df = df[df['Description'].notna()]
```

```python
#Try to remove most of the noise
phrase_list = {
  "essential functions:",
  "essential responsibilities:",
  "job description:",
  "role expectations:",
  "job role responsibilities:",
  "job specific responsibilities:",
  "what you will be doing:",
  "responsibilities include:",
  "responsibilities:",
  "job duties",
  "summary:"
}
for phrase in phrase_list:
  df['Description'].str.split(phrase, n=1).str.get(-1)
```

```python
# Lemmatizacion

import nltk
nltk.download('wordnet')

w_tokenizer = nltk.tokenize.WhitespaceTokenizer()
lemmatizer = nltk.stem.WordNetLemmatizer()

def lemmatize_text(text):
    return [lemmatizer.lemmatize(w) for w in w_tokenizer.tokenize(text)]

df['Description'] = df.Description.apply(lemmatize_text)
df['Description'] = df['Description'].apply(lambda x : " ".join(x))
```

```python
# Filter all elements where the ammount of words is less than thirty
df = df[df['Description'].apply(lambda x: len(x.split()) > 30)]
```
