```python
df = df[df['Description'].notna()]
words = " ".join(df['Description'])
#print(words)
```

```python
phrases_set = set()
time_set = set()
import time
# Importing the library for RAM usage
import psutil
```

```python
rake_object = RAKE.Rake(RAKE.SmartStopList())
for counter in range(10):
  start_time = time.time()
  k = rake_object.run(words, minCharacters = 5, maxWords = 5, minFrequency = 5)

  execution_time_seconds = round((time.time() - start_time), 2)
  print("--- %s seconds ---" % execution_time_seconds)
  time_set.add(execution_time_seconds)
  print(k[:20])

  #RAKE doesn’t originally print keywords in order of score. But it returns the score and the extracted keyphrases.
  # this will order them, thus we get the top 10 key phrases
  k.sort(key = lambda x: x[1], reverse = True)

  results = pd.DataFrame(k)
  for i in results.iloc[:,0]:
    phrases_set.add(i)

  # Getting % usage of virtual_memory ( 3rd field)
  print('RAM memory % used:', psutil.virtual_memory()[2])
  # Getting usage of virtual_memory in GB ( 4th field)
  print('RAM Used (GB):', psutil.virtual_memory()[3]/1000000000)
```
